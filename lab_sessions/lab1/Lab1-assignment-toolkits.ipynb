## [total points: 4] Exercise 1: NLTK
In this exercise, we use NLTK to apply **Part-of-speech (POS) tagging**, **Named Entity Recognition (NER)**, and **Constituency parsing**. The following code snippet already performs sentence splitting and tokenization. 

```python
import nltk
from nltk.tokenize import sent_tokenize
from nltk import word_tokenize
```

```python
sentences_nltk = sent_tokenize(text)
```

```python
tokens_per_sentence = []
for sentence_nltk in sentences_nltk:
    sent_tokens = word_tokenize(sentence_nltk)
    tokens_per_sentence.append(sent_tokens)
```

We will use lists to keep track of the output of the NLP tasks. We can hence inspect the output for each task using the index of the sentence.

```python
sent_id = 1
print('SENTENCE', sentences_nltk[sent_id])
print('TOKENS', tokens_per_sentence[sent_id])
```

### [point: 1] Exercise 1a: Part-of-speech (POS) tagging
Use `nltk.pos_tag` to perform part-of-speech tagging on each sentence.

Use `print` to **show** the output in the notebook (and hence also in the exported PDF!).

```python
pos_tags_per_sentence = []
for tokens in tokens_per_sentence:
    pos_tags = nltk.pos_tag(tokens)
    pos_tags_per_sentence.append(pos_tags)
    print()
```

```python
print(pos_tags_per_sentence)
```

### [point: 1] Exercise 1b: Named Entity Recognition (NER)
Use `nltk.chunk.ne_chunk` to perform Named Entity Recognition (NER) on each sentence.

Use `print` to **show** the output in the notebook (and hence also in the exported PDF!).

```python
from nltk.chunk import ne_chunk
ner_tags_per_sentence = []
for pos_tags in pos_tags_per_sentence:
    ner_tree = ne_chunk(pos_tags)
    ner_tags_per_sentence.append(ner_tree)
```

```python
print(ner_tags_per_sentence)
```

### [points: 2] Exercise 1c: Constituency parsing
Use the `nltk.RegexpParser` to perform constituency parsing on each sentence.

Use `print` to **show** the output in the notebook (and hence also in the exported PDF!).

```python
constituent_parser = nltk.RegexpParser('''
NP: {<DT>? <JJ>* <NN>*} # NP
P: {<IN>}           # Preposition
V: {<V.*>}          # Verb
PP: {<P> <NP>}      # PP -> P NP
VP: {<V> <NP|PP>*}  # VP -> V (NP|PP)*''')
```

```python
constituency_output_per_sentence = []
for pos_tags in pos_tags_per_sentence:
    parsed = constituent_parser.parse(pos_tags)
    constituency_output_per_sentence.append(parsed)


```

```python
print(constituency_output_per_sentence)
```

Augment the RegexpParser so that it also detects Named Entity Phrases (NEP), e.g., that it detects *Galaxy S III* and *Ice Cream Sandwich*

```python
constituent_parser_v2 = nltk.RegexpParser('''
NP: {<DT>? <JJ>* <NN>*} # NP
P: {<IN>}           # Preposition
V: {<V.*>}          # Verb
PP: {<P> <NP>}      # PP -> P NP
VP: {<V> <NP|PP>*}  # VP -> V (NP|PP)*
NEP: {<JJ|NNP><NNP|JJ|NN|CD>*<NNP|NN>}             # ???''')
```

```python
constituency_v2_output_per_sentence = []
for pos_tags in pos_tags_per_sentence:
    parsed = constituent_parser_v2.parse(pos_tags)
    constituency_v2_output_per_sentence.append(parsed)
```

We began by comparing sentence tokenization and part-of-speech (POS) tagging. Both NLTK and spaCy were able to split the sentence correctly without discrepancies, indicating that sentence splitting does not differ significantly for this specific case.

When comparing POS tagging, however, we observed some differences. NLTK uses the Penn Treebank tag set, labeling words like “Apple” as NNP (proper noun), “had” as VBD (verb, past tense), and “acted” as VBN (verb, past participle). On the other hand, spaCy uses a more modern tag set and provides both coarse-grained (pos_) and fine-grained (tag_) tags. For example, spaCy correctly identifies “Apple” as a proper noun (PROPN) and distinguishes auxiliary verbs like “had” with the tag AUX, which is more precise and informative than NLTK’s broader categories.

Overall, spaCy’s POS tagging output is more detailed, giving it an edge over NLTK in terms of accuracy.

### [points: 2] Exercise 3b: Named Entity Recognition (NER)
* Describe differences between the output from NLTK and spaCy for Named Entity Recognition. Which one do you think performs better?

We examined named entity recognition. NLTK offers NER capabilities through ne_chunk, but it requires prior POS tagging and represents recognized entities as chunks in a tree structure. In our analysis, NLTK was able to identify “Apple” as an organization, although the process was more manual.
In contrast, spaCy provided a straightforward and effective NER pipeline. Running the same sentence through spaCy’s en_core_web_sm model immediately identified “Apple” as an entity of type ORG (organization). SpaCy the clearly better and easier to use for NER. Its ability to perform high-quality named entity recognition with minimal configuration is a strong advantage over NLTK.

### [points: 2] Exercise 3c: Constituency/dependency parsing
Choose one sentence from the text and run constituency parsing using NLTK and dependency parsing using spaCy.
* describe briefly the difference between constituency parsing and dependency parsing
* describe differences between the output from NLTK and spaCy.

Finally, we looked at syntactic parsing, specifically the difference between constituency parsing (supported by NLTK) and dependency parsing (supported by spaCy). Constituency parsing breaks the sentence into nested phrase structures such as noun phrases (NP) and verb phrases (VP), typically visualized as a tree. Although NLTK can perform constituency parsing, doing so with real-world sentences often requires manually defining grammars, which can be complex and limited in scope.

SpaCy, on the other hand, provides built-in dependency parsing. Each token is connected to a head word, and grammatical relationships are clearly annotated. In our sentence, spaCy correctly identifies “Apple” as the subject of “stated”, and “acted” as part of a clausal complement to “stated”. This structured representation of dependencies makes syntactic analysis efficient and informative.

While NLTK is better suited for traditional constituency parsing (if extended properly), spaCy excels in dependency parsing thanks to its integrated and high-performance parser which gives better performance and reliability.
